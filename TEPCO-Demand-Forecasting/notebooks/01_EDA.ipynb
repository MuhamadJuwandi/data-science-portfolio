# EDA: TEPCO Electricity Demand Forecasting

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

## 2. Load Data

```python
df = pd.read_csv('../data/processed/final_dataset.csv')
df['Datetime'] = pd.to_datetime(df['Datetime'])
df.set_index('Datetime', inplace=True)
df.head()
```

## 3. Visualization

```python
plt.figure(figsize=(15, 6))
plt.plot(df['Demand'], label='Demand', alpha=0.5)
plt.title('TEPCO Demand History')
plt.legend()
plt.show()
```
