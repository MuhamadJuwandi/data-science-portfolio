# Feature Engineering for TEPCO Demand Forecasting

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import holidays
import matplotlib.pyplot as plt
```

## 2. Load Data

```python
df = pd.read_csv('../data/processed/final_dataset.csv')
df['Datetime'] = pd.to_datetime(df['Datetime'])
df = df.sort_values('Datetime')
df.head()
```

## 3. Create Time Features

```python
df['Hour'] = df['Datetime'].dt.hour
df['Month'] = df['Datetime'].dt.month
df['DayOfWeek'] = df['Datetime'].dt.dayofweek
df['IsWeekend'] = df['DayOfWeek'] >= 5

# Holiday
jp_holidays = holidays.JP()
df['IsHoliday'] = df['Datetime'].dt.date.apply(lambda x: x in jp_holidays)

df.head()
```

## 4. Lag & Rolling Features

```python
# Lags
df['Lag_1h'] = df['Demand'].shift(1)
df['Lag_24h'] = df['Demand'].shift(24)
df['Lag_168h'] = df['Demand'].shift(168)

# Rolling
df['RollingMean_24h'] = df['Demand'].shift(1).rolling(window=24).mean()
```

## 5. Save

```python
df = df.dropna()
df.to_csv('../data/processed/featured_dataset_notebook_ver.csv', index=False)
print("Saved.")
```
