Analyze hour of peak demand over years.

```python
df['Year'] = pd.to_datetime(df['Datetime']).dt.year
df['Hour'] = pd.to_datetime(df['Datetime']).dt.hour
peak_hours = df.loc[df.groupby(['Year', 'Date'])['Demand'].idxmax()]
peak_hour_dist = peak_hours.groupby('Year')['Hour'].mean()
print(peak_hour_dist)
peak_hour_dist.plot(kind='bar', title='Average Peak Hour by Year')
```
