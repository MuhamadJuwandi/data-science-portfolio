# Step 1: Data Understanding & Cleaning

**Objective:** Understand relational structure and clean the data.

**File:** `notebooks/1_data_cleaning.ipynb`

## Actions:
1. Read all 9 CSV files.
2. Merge and join tables.
3. Standardize dates, categories, and column names.
4. Translate product categories (Portuguese to English).
5. Handle missing values.
6. Save cleaned data.

```python
import pandas as pd
import os
```

```python
# Define paths
data_path = '../dataset'
output_file = '../dataset/cleaned_data.pkl'
```

```python
# Load Datasets
orders = pd.read_csv(os.path.join(data_path, 'olist_orders_dataset.csv'))
items = pd.read_csv(os.path.join(data_path, 'olist_order_items_dataset.csv'))
customers = pd.read_csv(os.path.join(data_path, 'olist_customers_dataset.csv'))
products = pd.read_csv(os.path.join(data_path, 'olist_products_dataset.csv'))
reviews = pd.read_csv(os.path.join(data_path, 'olist_order_reviews_dataset.csv'))
payments = pd.read_csv(os.path.join(data_path, 'olist_order_payments_dataset.csv'))
translation = pd.read_csv(os.path.join(data_path, 'product_category_name_translation.csv'))
```

```python
# Merge Datasets
print("Merging datasets...")
df = orders.merge(items, on='order_id', how='left')
df = df.merge(customers, on='customer_id', how='left')
df = df.merge(reviews, on='order_id', how='left')
df = df.merge(payments, on='order_id', how='left')
df = df.merge(products, on='product_id', how='left')
```

```python
# Translate Product Categories
product_category_map = dict(zip(translation['product_category_name'], translation['product_category_name_english']))
df['product_category_name_english'] = df['product_category_name'].map(product_category_map)
df['product_category_name_english'].fillna('unknown', inplace=True)
```

```python
# Standardize Dates
date_cols = ['order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date', 
             'order_delivered_customer_date', 'order_estimated_delivery_date', 'review_creation_date', 
             'review_answer_timestamp']

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')
```

```python
# Handle Missing Values
# Checking for missing values in critical columns
print(df.isnull().sum())
```

```python
# Save cleaned data
df.to_pickle(output_file)
print(f"Cleaned data saved to {output_file}")
```
